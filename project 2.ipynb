# Student Result Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("student_scores.csv")
print(df.head())
```

```python
df.describe()
```

```python
df.info()
```

```python
df = df.drop("Unnamed: 0" , axis = 1)
print(df.head())
```

# change weekly study hours column

```python
df["WklyStudyHours"]=df["WklyStudyHours"].str.replace("05-Oct","5-10")
df.head()
```

# gender distribution

```python
plt.figure(figsize=(5,5))
ax=sns.countplot(data=df,x="Gender")
ax.bar_label(ax.containers[0])
plt.title("gender distribution")
plt.savefig('plots/gender_distribution.png', bbox_inches='tight')
plt.show()
```

```python
#from the above chart we have analyzed that :
#the number of females in data is more than the number of males
```

```python
gb=df.groupby("ParentEduc").agg({"MathScore":'mean',"ReadingScore":"mean","WritingScore":"mean"})
print(gb)
```

```python
plt.figure(figsize=(4,4))
sns.heatmap(gb,annot=True)
plt.title("Relationship between parent's education and student's score")
plt.savefig('plots/parent_education_vs_student_score.png', bbox_inches='tight')
plt.show()
```

```python
#from the above chart we have concluded that the education of the parents have
#good impact on their scores
```

```python
gb1=df.groupby("ParentMaritalStatus").agg({"MathScore":"mean","ReadingScore":"mean","WritingScore":"mean"})
print(gb1)
```

```python
plt.figure(figsize=(4,4))
sns.heatmap(gb1,annot=True)
plt.title("Relationship between Marital status and student's score")
plt.savefig('plots/marital_status_vs_student_score.png', bbox_inches='tight')
plt.show()
```

```python
#from the above chart we have concluded that there is no/negligible impact on 
#student's score due to their parent's marital status
```

```python
sns.boxplot(data=df,x="MathScore")
plt.savefig('plots/math_score_distribution.png', bbox_inches='tight')
plt.show()
```

```python
sns.boxplot(data=df,x="ReadingScore")
plt.savefig('plots/reading_score_distribution.png', bbox_inches='tight')
plt.show()
```

```python
sns.boxplot(data=df,x="WritingScore")
plt.savefig('plots/writing_score_distribution.png', bbox_inches='tight')
plt.show()
```

# DIstribution of Ethnic Groups

```python
groupA = df.loc[(df['EthnicGroup']=="group A")].count()
groupB = df.loc[(df['EthnicGroup']=="group B")].count()
groupC = df.loc[(df['EthnicGroup']=="group C")].count()
groupD = df.loc[(df['EthnicGroup']=="group D")].count()
groupE = df.loc[(df['EthnicGroup']=="group E")].count()

l=["group A","group B","group C","group D","group E"]
mlist=[groupA["EthnicGroup"],groupB["EthnicGroup"],groupC["EthnicGroup"],groupD["EthnicGroup"],groupE["EthnicGroup"]]
plt.pie(mlist,labels=l,autopct="%1.2f%%")
plt.title("Distribution of Ethnic Groups")
plt.savefig('plots/ethnic_group_distribution.png', bbox_inches='tight')
plt.show()

```

```python
ax=sns.countplot(data=df,x='EthnicGroup')
ax.bar_label(ax.containers[0])
plt.title("EthnicGroup vs count")
plt.savefig('plots/ethnic_group_count.png', bbox_inches='tight')
```
